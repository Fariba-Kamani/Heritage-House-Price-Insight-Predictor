# DataFrame Summary
* Display column names, data types, and unique values
    * Data Types: int64 (13 columns), float64 (7 columns), object (4 columns)
    * Missing Data: 2ndFlrSF, BedroomAbvGr, BsmtExposure, BsmtFinType1, EnclosedPorch, GarageFinish, GarageYrBlt, LotFrontage, MasVnrArea , and WoodDeckSF
    * Categorical columns such as GarageFinish, KitchenQual, and BsmtExposure have low cardinality (3–6 unique values), confirming they are suitable for encoding.
    * Continuous numerical columns like LotArea, GrLivArea, and SalePrice have high cardinality (hundreds of unique values), indicating they are continuous variables appropriate for regression analysis.
    * Features like OverallQual and OverallCond are numerical but have ordinal characteristics (e.g., quality ratings from 1–10).

```python
# Basic structure and types
price_records_df.info()

# Number of unique values per column
price_records_df.nunique().sort_values()
```

* Identify variable types: numerical, categorical, datetime

```python
# Identify numerical columns
numerical_cols = price_records_df.select_dtypes(
    include=['int64', 'float64']).columns.tolist()

# Identify categorical columns
categorical_cols = price_records_df.select_dtypes(
    include='object').columns.tolist()

# Identify datetime columns
datetime_cols = price_records_df.select_dtypes(
    include='datetime').columns.tolist()

# Print results
print("Numerical columns:", numerical_cols)
print("Categorical columns:", categorical_cols)
print("Datetime columns:", datetime_cols)
```
